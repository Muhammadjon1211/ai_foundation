```python
import numpy as np
```

```python
arr = np.array([1,2,3,4])
print(arr)
```

```python
arr2 = np.array([[1,2,3,4], [6,7,8,9]])
print(arr2)
```

```python
#dtype
a = np.array([1,2,3])
print(a.dtype)
```

```python
#reshape
a = np.array([1,2,3,4,5,6])
b = a.reshape(2,3)
print(b)
```

```python
#astype
a = np.array([1,2,3])
b = a.astype(float)
print(b)
```

```python
#np.zeros(), np.ones(), => dummy data
z = np.zeros((2,3))
o = np.ones((2,3))
print(z)
print(o)
```

```python
#ndim
a = np.array([[1,2], [3,4]])
print(a.ndim)
```

```python
#arange()
a = np.arange(1,7)
print(a)
```

```python
#sum, mean, max, min
a = np.array([[1,2], [3,4]])
print(a.sum())
print(a.mean())
print(a.max())
print(a.min())
```

```python
#flatten()
a = np.array([[1,2], [3,4]])
b = a.flatten()
print(b)
```

```python
#T
a = np.array([[1,2], [3,4]])
b = a.T
print(b)
```

```python
#dot()@
a = np.array([[1,2], [3,4]])
b = np.array([[2,0], [1,2]])
print(a.dot(b))
```

```python
# random .rand()
a = np.random.rand(2,3)
print(a)
```
